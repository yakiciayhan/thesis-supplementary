## Inter-Rater Reliability Analysis and Consensus Resolution

Two independent raters initially scored 32 evaluation units across four dimensions (D1, D2, D3, D4) using an ordinal rating scale. Inter-rater reliability was calculated using **Cohen's Kappa** based on these pre-discussion independent scores.

The overall initial performance across all dimensions pooled yielded a kappa of **0.8390** (118/128 exact agreement), which falls within the *Almost Perfect* range according to the Landis & Koch (1977) classification.

When analyzed per dimension, **D4** demonstrated near-perfect reliability with a kappa of **0.93** (31/32 exact agreement). **D2** yielded a *Substantial* kappa of **0.65** (31/32 exact agreement), while **D3** showed *Moderate* agreement with a kappa of **0.54** (26/32 exact agreement).

**Methodological Note on D1:**
For **D1**, the exact agreement between raters was notably high (30 out of 32 cases). However, the calculated statistical kappa value is **0.00**. This is a recognized statistical phenomenon known as the *Kappa prevalence/base-rate paradox* (Feinstein & Cicchetti, 1990). Because almost all cases fell into a single dominant scoring category (a score of 2), the variance was extremely low. In highly skewed marginal distributions like this, the kappa statistic artificially collapses to zero despite highly consistent actual agreement. Therefore, the exact agreement metric (93.75%) is utilized as the primary reflection of inter-rater reliability for this specific dimension instead of the statistical kappa.
