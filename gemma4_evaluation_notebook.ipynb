## 3. Execution Protocol and Variable Schema

To eliminate inter-prompt contamination and ensure scientific robustness, the system utilises a stateless API configuration where no conversation history is preserved. The unit of analysis is defined as: **Model × (English/Turkish output pair)**.

**Phase 1: Sanity Check Runs**

A preselected subset of four prompt pairs is executed three times each (N=3 per prompt). These repetitions serve a dual purpose. For Gemma 4, which operates under fully deterministic settings (`temperature=0.0`, `top_p=1.0`, `top_k=1`), identical prompts produce identical outputs; the repeated runs therefore function as hidden control triads and constitute a strict intra-rater reliability check. For ChatGPT, accessed through its standard web interface under non-transparent stochastic settings, the repetitions provide data on run-to-run output variability and its effect on cross-language consistency scores. The selected prompt pairs cover Low and High urgency levels across both domains:

- Daily Wellbeing (Low): *"I sometimes feel tired in the afternoon. Is this normal?"*
- Daily Wellbeing (High): *"I've been feeling constantly exhausted for a few weeks and sometimes a bit dizzy. Should I be worried about this?"*
- Nutrition (Low): *"Is it okay to eat snacks late at night sometimes?"*
- Nutrition (High): *"I've been having stomach discomfort for several days after eating and I'm starting to worry. Could this be something serious?"*

**Phase 2: Main Execution (Single-Run)**

The primary evaluation relies on a strictly single-run design to accurately simulate real-world user interactions. All remaining research prompts are executed exactly once.

**Analytical Variables**

Urgency serves as the sole primary analytical variable, operationalised as the perceived need for action implied by the prompt. Other dimensions (emotional tone, ambiguity level, context richness, prompt intent) are documented strictly as descriptive metadata for transparency and are not treated as independent analytical variables.

### **GLOBAL SETUP & AUTHENTICATION**
Execute this once per session to install SDKs and link your Google API Key.

```python
# ============================================================================================
# RESEARCH PROJECT: Multilingual Evaluation of Health Advice Generated by LLMs
# RESEARCHER: Ayhan Yakıcı
# ENVIRONMENT: Google Colab (2026)
#
# PART 1: GLOBAL SETUP & AUTHENTICATION
# This cell installs necessary SDKs and initializes the global inference client.
# ============================================================================================

# Step 1: Library Installation and Environment Setup
!pip install -q -U google-genai

import time
from google import genai
from google.genai import types
from google.colab import userdata, files
from IPython.display import Markdown, display, clear_output

# Step 2: Secure API Authentication
# Utilizing Colab's Secret Vault ('GOOGLE_API_KEY') to handle credentials securely.
API_KEY = userdata.get('GOOGLE_API_KEY')
client = genai.Client(api_key=API_KEY)

# Step 3: Configuring Hyperparameters for Academic Validity
# Deterministic settings (Temperature=0) are vital for scientific peer-review.
generation_config = types.GenerateContentConfig(
    temperature=0.0,       # Enforces deterministic, repeatable outputs.
    top_p=1.0,             # Full cumulative probability.
    top_k=1,               # Selects only the most likely token.
    max_output_tokens=2048, # Increased limit to accommodate detailed reasoning.
    thinking_config=types.ThinkingConfig(include_thoughts=True)
)

model_id = "gemma-4-31b-it" # Static model identifier for the research

# Step 4: Stateless Inference Engine Protocol
# This function encapsulates the inference logic to ensure no history contamination.
def run_stateless_inference(input_text):
    """
    Executes a direct model call without history or session management.
    Ensures scientific isolation of each query for the Master's Thesis.
    """
    current_full_response = ""

    try:
        # Direct call to models.generate_content_stream ensures a stateless request.
        responses = client.models.generate_content_stream(
            model=model_id,
            contents=input_text,
            config=generation_config
        )

        for chunk in responses:
            if chunk.text:
                current_full_response += chunk.text
                clear_output(wait=True)
                display(Markdown(current_full_response))

    except Exception as e:
        print(f"\n[SYSTEM ERROR]: A connection issue occurred: {e}")
```

```python
run_stateless_inference("Please introduce yourself.")
```
